#  Cleaned Company Employee Dataset Using Pandas-IDRA


**Objective:** Clean a messy company employee dataset using Pandas. The analysis identifies missing values, inconsistent categorical entries, incorrect data types, and duplicate records. Appropriate cleaning techniques are applied, the dataset is verified after cleaning, compared before and after, and exported as a cleaned CSV file.

---
## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
```

## 2. Upload and Load the Messy Dataset

```python
from google.colab import files
uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print('Loaded File:', file_name)
display(df.head())
```

---
# 3. Initial Dataset Inspection

```python
print('Dataset Shape:', df.shape)
print('Number of Rows:', df.shape[0])
print('Number of Columns:', df.shape[1])
print('\nColumn Names:')
print(df.columns.tolist())

display(df.head())
display(df.tail())
```

```python
# Inspect data types and non-null values
df.info()
```

```python
# Descriptive statistics
display(df.describe())
display(df.describe(include='object'))
```

---
# 4. Before-Cleaning Summary

```python
before_summary = {
    'Rows': df.shape[0],
    'Columns': df.shape[1],
    'Total Missing Values': int(df.isnull().sum().sum()),
    'Duplicate Rows': int(df.duplicated().sum())
}

before_summary_df = pd.DataFrame([before_summary])
display(before_summary_df)
```

---
# 5. Identify and Quantify Missing Values

```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().mean() * 100).round(2)

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage (%)': missing_percentage
})

display(missing_summary)
```

### Missing Value Analysis
We inspect each column before choosing a cleaning method. Numerical columns will use **median imputation** because the median is less affected by extreme values. Categorical columns will use **mode imputation** because the most frequent category is a reasonable replacement.

---
# 6. Identify Duplicate Records

```python
duplicate_count = df.duplicated().sum()
print('Number of Duplicate Rows:', duplicate_count)

if duplicate_count > 0:
    print('\nSample Duplicate Records:')
    display(df[df.duplicated(keep=False)].head(20))
```

---
# 7. Identify Inconsistent Entries
Categorical columns are checked using `unique()` and `value_counts()` to detect inconsistent capitalization and unwanted spaces.

```python
categorical_columns = ['Department', 'Gender', 'City', 'Work_Mode']

for column in categorical_columns:
    print(f'\n{column}:')
    print('Unique Values:', df[column].dropna().unique())
    display(df[column].value_counts(dropna=False))
```

### Problems Identified
The inspection reveals examples such as:
- `Engineering`, `ENGINEERING`, and `engineering`
- `Female`, `female`, and `MALE`
- `Delhi`, `DELHI`, `delhi`, and `Delhi `
- `Remote`, `REMOTE`, and `remote`

These entries represent the same categories but are stored inconsistently.

---
# 8. Create a Copy for Cleaning

```python
# Keep the original dataset unchanged for comparison
clean_df = df.copy()
```

# 9. Clean Inconsistent Categorical Values

```python
# Remove leading and trailing spaces
for column in ['Department', 'Gender', 'City', 'Work_Mode']:
    clean_df[column] = clean_df[column].astype('string').str.strip()

# Standardize capitalization
clean_df['Department'] = clean_df['Department'].str.title()
clean_df['Gender'] = clean_df['Gender'].str.title()
clean_df['City'] = clean_df['City'].str.title()
clean_df['Work_Mode'] = clean_df['Work_Mode'].str.title()

for column in ['Department', 'Gender', 'City', 'Work_Mode']:
    print(f'\nCleaned {column} values:')
    print(clean_df[column].dropna().unique())
```

---
# 10. Handle Missing Numerical Values Using Median Imputation

```python
numerical_columns = ['Age', 'Annual_Salary', 'Experience_Years', 'Performance_Score']

for column in numerical_columns:
    median_value = clean_df[column].median()
    print(f'{column} - Median used for imputation: {median_value}')
    clean_df[column] = clean_df[column].fillna(median_value)
```

# 11. Handle Missing Categorical Values Using Mode Imputation

```python
categorical_columns = ['Department', 'Gender', 'City', 'Work_Mode']

for column in categorical_columns:
    mode_value = clean_df[column].mode()[0]
    print(f'{column} - Mode used for imputation: {mode_value}')
    clean_df[column] = clean_df[column].fillna(mode_value)
```

### Note on Other Missing-Value Methods
Pandas also provides `dropna()` and forward filling (`ffill()`):
- `dropna()` is useful when rows with missing values are unsuitable for analysis or contain too much missing information.
- `ffill()` is useful when observations have a meaningful sequential order, such as time-series data.

For this employee dataset, median and mode imputation are more appropriate because dropping rows would unnecessarily reduce the dataset, and forward filling could incorrectly copy one employee's information into another employee's record.

---
# 12. Correct Data Types

```python
# Convert Joining_Date from object/string to datetime
clean_df['Joining_Date'] = pd.to_datetime(clean_df['Joining_Date'], errors='coerce')

print(clean_df.dtypes)
```

# 13. Remove Duplicate Records

```python
rows_before_duplicates = len(clean_df)
clean_df = clean_df.drop_duplicates().copy()
rows_after_duplicates = len(clean_df)

print('Rows before removing duplicates:', rows_before_duplicates)
print('Rows after removing duplicates:', rows_after_duplicates)
print('Duplicates removed:', rows_before_duplicates - rows_after_duplicates)
```

---
# 14. Verify the Cleaned Dataset

```python
print('Missing Values After Cleaning:')
display(clean_df.isnull().sum())

print('Duplicate Rows After Cleaning:', clean_df.duplicated().sum())

print('\nCleaned Dataset Information:')
clean_df.info()
```

# 15. Compare Dataset Before and After Cleaning

```python
after_summary = {
    'Rows': clean_df.shape[0],
    'Columns': clean_df.shape[1],
    'Total Missing Values': int(clean_df.isnull().sum().sum()),
    'Duplicate Rows': int(clean_df.duplicated().sum())
}

comparison = pd.DataFrame({
    'Before Cleaning': before_summary,
    'After Cleaning': after_summary
})

display(comparison)
```

```python
output_file = 'Cleaned_Company_Employee_Dataset.csv'
clean_df.to_csv(output_file, index=False)

print(f'Cleaned dataset successfully saved as: {output_file}')

# Download the CSV file
files.download(output_file)
```
